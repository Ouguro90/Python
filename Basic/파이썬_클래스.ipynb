# 클래스 (class) 와 객체 (object)

## 클래스

관련있는 변수와 기능을 묶어 하나의 설계도로 표현하는 방식  

어떤 객체가 어떤 속성과 동작을 가질지 정의하는 틀

```python
class Dog:
  def __init__(self, name):
    self.name = name
  
  def bark(self):
    print(f'{self.name}가 멍멍 짖어요')
```

Dog 라는 클래스는 '개' 라는 존재에 대한 설계도이다  
이름을 저장하고 짖는 동작을 가지고 있다

### 구성 요소

#### __init__() 생성자

객체를 만들 때 자동으로 실행되는 함수,  

 객체 초기화 용도로 사용된다

```python
class Dog:
  def __init__(self, name, breed):
    self.name = name
    self.breed = breed
```

#### self

현재 객체에서 자기 자신을 가리킴  

메서드에서 반드시 첫 인자로 사용함  

객체마다 고유하게 속성 저장 가능

## 객체

클래스를 이용해 만든 실제 실체  

클래스는 틀이고, 객체는 그 틀로 만든 실제 물건

```python
my_dog = Dog('초코')  # Dog 클래스의 객체 생성
my_dog.bark()

# 초코가 멍멍 짖어요
```

my_dog가 Dog 클래스의 객체이다

## 상속

부모 클래스의 기능을 자식 클래스가 물려 받는것

코드의 재사용이 가능해지고, 공통된 기능을 한곳에서 정의할 수 있어 편해진다

```python
class Animal:  # 부모 클래스
    def speak(self):
        print("소리를 냅니다.")

class Dog(Animal):  # 자식 클래스
    def bark(self):
        print("멍멍!")

dog = Dog()
dog.speak()  # 부모의 메서드 사용
dog.bark()   # 자신의 메서드 사용


소리를 냅니다.
멍멍!
```

Dog 클래스는 Animal 클래스를 상속받았기 떄문에, speak()를 그대로 사용 가능

## 메서드 오버라이딩

오버라이딩은 부모 클래스에서 정의된 메서드를 자식 클래스에서 재정의 하는것  

기능을 자식 클래스에 맞게 변경하고 싶을 때 사용

```python
class Animal:
    def speak(self):
        print("동물이 소리를 냅니다.")

class Cat(Animal):
    def speak(self):  # 오버라이딩!
        print("야옹!")

cat = Cat()
cat.speak()

야옹!
```

부모의 speak()는 "동물이 소리를 냅니다"였지만, 자식 클래스 Cat에서 오버라이딩했기 때문에 "야옹!"으로 동작이 바뀌었다
