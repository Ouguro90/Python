# 파일 생성하기

- open(파일_이름, 파일_열기_모드) 를 사용한다
- 파일 열기 모드에는 r, w, a가 있다
- 쓰기 모드('w')로 열면 파일이 없으면 새로 만들고, 있으면 내용이 지워진다

```python
#f = open(파일 이름, 파일 열기 모드(w: write, 쓰기 모드로 열기))
f = open('새파일.txt', 'w')
f.close()
```

- 경로를 적을 때는 / 또는 \ 또는 r'' 를 사용해야 한다

```python
# 다른 디렉토리에 있는 파일을 열 때에는 경로를 직접 지정
f = open('새폴더//새파일.txt', 'w')
f.close()
```

# 파일을 쓰기 모드로 열어 내용 쓰기

```python
# 1. 동일 디렉토리에 있는 새파일.txt 쓰기 모드로 열기
f = open('새파일.txt', 'w')

# 반복문 실행 (1~10)
for i in range(1, 11):
  # data에 값 넣기
  data = f'{i}번째 줄입니다 \n'
  # 파일에 data 내용 쓰기
  f.write(data)
  print()

f.write('문자열 파일에 쓰기')
f.close()
```

# 파일 읽기

## read() 함수 사용하기

- read() 함수는 전체 내용을 문자열로 읽는다

```python
f = open('새파일.txt', 'r')
content = f.read()
print(content)
f.close()
```

## readline() 함수 사용하기

- readline() 함수는 파일의 내용 중 한줄만 읽는다

```python
f = open('새파일.txt', 'r')
content = f.readline()
print(content)
f.close()
```

## readlines() 함수 사용하기

- readlines() 함수는 모든 줄을 읽고 리스트로 표시한다

```python
f = open('새파일.txt', 'r')
content = f.readlines()
print(content)
f.close()
```

# 파일에 새로운 내용 추가하기

- 'a' 모드로 열면 기존 내용 뒤에 추가된다

```python
f = open('새파일.txt', 'a')
for i in range(11,20):
  data = f'{i}번째 줄입니다 \n'
  f.write(data)

f.close()
```

# with 문과 함께 사용하기

- with open()을 사용하면 close() 를 따로 호출하지 않아도 된다

```python
with open('foo.txt', 'w') as f:
  f.write('Life is too short, you need python')
```
