# Q1 홀수, 짝수 판별하기

주어진 자연수가 홀수인지, 짝수인지 판별해 주는 함수 is_odd를 작성  

is_odd 함수는 홀수이면 True, 짝수이면 False를 리턴해야 한다

````python
def is_odd(number):
  if number % 2 == 1:
    return True
  else:
    return False
````

# Q2 모든 입력의 평균값 구하기

입력으로 들어오는 모든 수의 평균값을 계산해 주는 함수를 작성  

단, 입력으로 들어오는 수의 개수는 정해져 있지 않다

````python
def avg_numbers(*args):
  result = 0
  for i in args:
    result += i
  return result / len(args)

avg_numbers(1, 2)
````

다음은 2개의 숫자를 입력받아 더한 후에 리턴하는 프로그램이다

# Q5 프로그램 오류 수정 2

다음은 파일 (text.txt)에 'Life is too short' 문자열을 저장한 후 다시 그 파일을 읽어  
출력하는 프로그램이다

```python
f1 = open('test.txt', 'w')
f1.write('Life is too short')

f2 = open('test.txt', 'r')
print(f2.read())
```


이 프로그램은 'Life is too short' 라는 문장을 출력하지 않는다  
예상한 값을 출력하도록 프로그램을 수정

````python
f1 = open('test.txt', 'w')
f1.write('Life is too short')
f1.close() # close()로 닫아줘야 함

f2 = open('test.txt', 'r')
print(f2.read())
````

# Q6 사용자 입력 저장

사용자의 입력을 파일(text.txt)에 저장하는 프로그램을 작성  

단, 프로그램을 다시 실행하더라도 기존에 작성한 내용을 유지하고 새로 입력한 내용을 추가

# Q7 파일의 문자열 바꾸기

다음과 같은 내용을 지닌 text.txt가 있다  

이 파일의 내용 중 'java'라는 문자열을 'python' 으로 바꾸어 저장  

```python
Life is too short

you need java
```

````python
f = open('test.txt', 'r')
body = f.read()
f.close()

body = body.replace('java', 'python')

f = open('test.txt', 'w')
f.write(body)
f.close()
````

# Q8 입력값을 모두 더해 출력

다음과 같이 실행할 때 입력값을 모두 더해 출력하는 스크립트(C:\doit\myargv.py)를 작성  


````python
import sys
````

````python
numbers = sys.argv[1:]

result = 0

for number in numbers:
  result += int(number)

print(result)
````
