# API란 무엇인가?

API는 서로 다른 소프트웨어 시스템 간에 정보를 주고받거나 기능을 사용할 수 있도록 도와주는 서비스  

쉽게 말하면, "요청하면 필요한 데이터를 보내주는 자동화된 시스템" 이다

# 고양이 API

```python
import requests
from IPython.display import Image,display

url = 'https://api.thecatapi.com/v1/images/search'
response = requests.get(url)
data = response.json()
cat_url = data[0]['url']

print('오늘의 고양이 사진:')
display(Image(url=cat_url))
```

# 날씨 정보 API (Open-Meteo API)

대표 예제 코드

```python
import requests
from datetime import datetime

latitude = 37.5665
longitude = 126.9780
today = '2025-04-14'

url = (
    f'https://api.open-meteo.com/v1/forecast?'
    f'latitude={latitude}&longitude={longitude}'  # 수정됨
    f'&daily=temperature_2m_max,temperature_2m_min'
    f'&timezone=Asia%2FSeoul'
)

response = requests.get(url)
data = response.json()

dates = data['daily']['time']
max_temps = data['daily']['temperature_2m_max']
min_temps = data['daily']['temperature_2m_min']

for date in dates:
    if date == today:
        i = dates.index(date)
        print(f'오늘({today}) 서울 날씨')
        print(f'- 최고 기온: {max_temps[i]}℃')
        print(f'- 최저 기온: {min_temps[i]}℃')
        break
```

# 오늘의 명언 API

```python
import requests
import json

url = 'https://zenquotes.io/api/random'
response = requests.get(url)
data = response.json()

print(json.dumps(data, indent=3))

quote = data[0]['q']
author = data[0]['a']

print('오늘의 명언:')
print(f'"{quote}" - {author}')
```

# 아침 루틴 만들기 (routine 메소드)

- 고양이 사진 확인
- 날씨 확인
- 명언 확인
- 추가 기능 제공 (자율)

```python
import requests
import json
from datetime import datetime
from IPython.display import Image,display
```

```python
### 루틴 클래스
class Routine:

  def __init__(self, latitude, longitude):
    self.latitude = latitude
    self.longitude = longitude
    self.today = datetime.today().strftime('%Y-%m-%d')

  ###### 고양이 사진
  def catphoto(self):
    url = 'https://api.thecatapi.com/v1/images/search'
    response = requests.get(url)
    data = response.json()
    cat_url = data[0]['url']

    print('오늘의 고양이 사진:')
    display(Image(url=cat_url))

  ###### 오늘의 날씨 (위도, 경도 입력)
  def today_weather(self):

    url = (
    f'https://api.open-meteo.com/v1/forecast?'
    f'latitude={self.latitude}&longitude={self.longitude}'
    f'&daily=temperature_2m_max,temperature_2m_min'
    f'&timezone=Asia%2FSeoul'
    )
    response = requests.get(url)
    data = response.json()

    dates = data['daily']['time']
    max_temps = data['daily']['temperature_2m_max']
    min_temps = data['daily']['temperature_2m_min']

    for date in dates:
      if date == self.today:
        i = dates.index(date)
        print(f'오늘({self.today}) 서울의 날씨')
        print(f'- 최고 기온: {max_temps[i]}℃')
        print(f'- 최저 기온: {min_temps[i]}℃')
        break

  ###### 오늘의 명언
  def today_quote(self):
    url = 'https://zenquotes.io/api/random'
    response = requests.get(url)
    data = response.json()

    quote = data[0]['q']
    author = data[0]['a']

    print('오늘의 명언')
    print(f'"{quote}" - {author}')

  ###### 오늘의 영단어
  def random_words(self):
    url = 'https://random-word-api.herokuapp.com/word?number=1'
    response = requests.get(url)
    word = response.json()

    print(f"오늘의 영단어: {word}")

  ### 루틴 메소드
  def routine(self):
    self.catphoto()
    self.today_weather()
    self.today_quote()
    self.random_words()
```

```python
### 루틴 호출
r = Routine(latitude=37.5665, longitude=126.9780)
r.routine()
```
